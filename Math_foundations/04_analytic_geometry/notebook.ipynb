# Analytic Geometry — Hands-on

## 1. Distance metrics

```python
import numpy as np

p1 = np.array([1, 2])
p2 = np.array([4, 6])

euclidean = np.linalg.norm(p1 - p2)
cosine_sim = np.dot(p1, p2) / (np.linalg.norm(p1) * np.linalg.norm(p2))

print("Euclidean distance:", euclidean)
print("Cosine similarity:", cosine_sim)
```

## 2. Visualizing a separating hyperplane (2D line)

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)
class_a = np.random.randn(20, 2) + np.array([2, 2])
class_b = np.random.randn(20, 2) + np.array([-2, -2])

plt.scatter(class_a[:,0], class_a[:,1], label='Class A')
plt.scatter(class_b[:,0], class_b[:,1], label='Class B')

xs = np.linspace(-5, 5, 10)
ys = -xs  # simple separating line y = -x
plt.plot(xs, ys, 'k--', label='hyperplane')
plt.legend()
plt.title("A hyperplane separating two classes")
plt.show()
```

## 3. L1 vs L2 norm penalty effect

```python
w = np.array([3.0, -4.0])

l1_norm = np.sum(np.abs(w))
l2_norm = np.sqrt(np.sum(w**2))

print("L1 norm:", l1_norm)
print("L2 norm:", l2_norm)
```

## 4. Angle between vectors

```python
theta_rad = np.arccos(cosine_sim)
theta_deg = np.degrees(theta_rad)
print("Angle between p1 and p2:", theta_deg, "degrees")
```

## 5. Manhattan (L1) distance vs Euclidean (L2) distance

```python
manhattan = np.sum(np.abs(p1 - p2))
print("Manhattan distance:", manhattan)
print("Euclidean distance:", euclidean)

# Manhattan distance matters for high-dimensional / grid-like data
# (e.g. city block distance, sparse feature spaces)
```

## 6. Vector projection

```python
def project(a, b):
    """Project vector a onto vector b"""
    return (np.dot(a, b) / np.dot(b, b)) * b

a = np.array([3, 4])
b = np.array([5, 0])  # projecting onto the x-axis

proj = project(a, b)
print("Projection of a onto b:", proj)

fig, ax = plt.subplots()
ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='a')
ax.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='green', label='b')
ax.quiver(0, 0, proj[0], proj[1], angles='xy', scale_units='xy', scale=1, color='red', label='projection of a onto b')
ax.set_xlim(-1, 6)
ax.set_ylim(-1, 6)
ax.legend()
ax.set_title("Vector projection")
plt.show()
```

## 7. KNN-style distance comparison across multiple points

```python
points = np.array([[1, 1], [2, 3], [5, 5], [-1, -2]])
query = np.array([0, 0])

distances = np.linalg.norm(points - query, axis=1)
nearest_idx = np.argmin(distances)

print("Distances from query point:", distances)
print("Nearest point:", points[nearest_idx])
```
