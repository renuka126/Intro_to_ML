# Calculus — Hands-on

## 1. Symbolic differentiation with sympy

```python
import sympy as sp

x = sp.Symbol('x')
f = x**2 + 3*x + 5
f_prime = sp.diff(f, x)

print("f(x) =", f)
print("f'(x) =", f_prime)
```

## 2. Gradient descent from scratch on y = x^2

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x**2

def grad_f(x):
    return 2*x

x = 8.0  # starting point
learning_rate = 0.1
history = [x]

for step in range(30):
    x = x - learning_rate * grad_f(x)
    history.append(x)

print("Final x:", x, " f(x):", f(x))

xs = np.linspace(-10, 10, 200)
plt.plot(xs, f(xs), label="y = x^2")
plt.plot(history, [f(h) for h in history], 'ro-', label="descent path")
plt.legend()
plt.title("Gradient descent converging to the minimum")
plt.show()
```

## 3. Learning rate: too high vs too low vs just right

```python
def run_gd(lr, steps=20, start=8.0):
    x = start
    path = [x]
    for _ in range(steps):
        x = x - lr * grad_f(x)
        path.append(x)
    return path

for lr, label in [(0.01, "too low"), (0.1, "good"), (1.05, "too high")]:
    path = run_gd(lr)
    print(f"lr={lr} ({label}): final x = {path[-1]:.3f}")
```
