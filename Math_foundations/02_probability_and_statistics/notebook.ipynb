# Probability & Statistics — Hands-on

## 1. Distributions

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
normal_samples = np.random.normal(loc=0, scale=1, size=1000)
bernoulli_samples = np.random.binomial(n=1, p=0.3, size=1000)
binomial_samples = np.random.binomial(n=10, p=0.5, size=1000)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].hist(normal_samples, bins=30); axes[0].set_title("Normal")
axes[1].hist(bernoulli_samples, bins=2); axes[1].set_title("Bernoulli (p=0.3)")
axes[2].hist(binomial_samples, bins=10); axes[2].set_title("Binomial (n=10, p=0.5)")
plt.show()
```

## 2. Bayes' theorem — simple spam filter example

```python
# P(spam) prior
p_spam = 0.4
p_not_spam = 0.6

# P(word "free" | spam) and P(word "free" | not spam)
p_free_given_spam = 0.7
p_free_given_not_spam = 0.1

# P(word "free") total
p_free = p_free_given_spam * p_spam + p_free_given_not_spam * p_not_spam

# Bayes: P(spam | "free")
p_spam_given_free = (p_free_given_spam * p_spam) / p_free
print(f"P(spam | contains 'free') = {p_spam_given_free:.3f}")
```

## 3. Mean, variance, and normalization

```python
data = np.array([10, 200, 15, 300, 12, 250])

mean = data.mean()
std = data.std()
normalized = (data - mean) / std

print("Original:", data)
print("Mean:", mean, "Std:", std)
print("Normalized:", normalized)
```
