# Linear Algebra — Hands-on
Working through vectors, matrices, and transformations with numpy.

## 1. Vectors and a feature matrix

```python
import numpy as np

# A tiny "dataset": 3 samples, 2 features each
X = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])
print("Feature matrix X:\n", X)
print("Shape:", X.shape)
```

## 2. Dot product and cosine similarity

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

dot = np.dot(a, b)
cos_sim = dot / (np.linalg.norm(a) * np.linalg.norm(b))

print("Dot product:", dot)
print("Cosine similarity:", cos_sim)
```

## 3. Matrix multiplication (manual vs numpy)

```python
W = np.array([[0.5, -0.2], [0.1, 0.4]])
x = np.array([1.0, 2.0])

# Manual
manual = np.array([
    W[0,0]*x[0] + W[0,1]*x[1],
    W[1,0]*x[0] + W[1,1]*x[1]
])

# numpy
auto = W @ x

print("Manual:", manual)
print("numpy:", auto)
```

## 4. Transpose and rank

```python
print("X transpose:\n", X.T)
print("Rank of X:", np.linalg.matrix_rank(X))
```

## 5. Visualizing a 2D transformation

```python
import matplotlib.pyplot as plt

theta = np.pi / 4  # 45 degree rotation
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])

v = np.array([1, 0])
v_rotated = R @ v

fig, ax = plt.subplots()
ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='blue', label='original')
ax.quiver(0, 0, v_rotated[0], v_rotated[1], angles='xy', scale_units='xy', scale=1, color='red', label='rotated')
ax.set_xlim(-2, 2)
ax.set_ylim(-2, 2)
ax.axhline(0, color='gray', lw=0.5)
ax.axvline(0, color='gray', lw=0.5)
ax.legend()
ax.set_title("Vector rotation by 45 degrees")
plt.show()
```
